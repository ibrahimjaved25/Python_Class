## Lambda Functions

> lambda arguments: expression

> Using map() with Lambda
> 
> Using filter() with Lambda

> Using reduce() with Lambda

> Lambda in Sorting Functions

```python
add = lambda a, b: a + b
print(add(5, 3))
```

```python
square = lambda x: x**2
print(square(3))
```

```python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

```python
num = [10,9,8,7,6]
result = list(map(lambda x:x*2, num))
print(result)
```

```python
prices = [10000,20000,30000,40000,50000]

discount = list(map(lambda x:x*0.90 ,prices ))
print(discount)
```

```python
prices = [10000,20000,30000,40000,50000]

selected = filter(lambda x:x<30000 ,prices )

discount = list(map(lambda x: x*0.90,selected))

print(discount)
```

```python
prices = [10000, 20000, 30000, 40000, 50000]

selected = filter(lambda x: x < 30000, prices)

discount = list(map(lambda x: x * 0.90, selected))

print(discount)
```

```python
# from functools import reduce

numbers = [10, 20, 30, 40]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

```python
students = [("Alice", 85), ("Bob", 78), ("Charlie", 92)]
students.sort(key=lambda student: student[1]) # Sort by score
print(students)
```

```python
num = [1,6,3,9,2,0]

result= sorted(num, key=lambda x: x)

print(result)
```
