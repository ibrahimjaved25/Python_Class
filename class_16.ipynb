```python
def display_users(data):
    print("User and their connection\n")
    for user in data['users']:
        print(f" {user['id']}: {user['name']} is friends with {user['friends']} and liked pages {user['liked_pages']} ")
    print("\npages information")
    for page in data['pages']:
        print(f" {page['id']}: and name is {page['name']}" )
 
display_users(data)
```
