# Object Oriented Programming

```python
class student:

    # Constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def study(self):   # this is method 
        print(f"my name is {self.name} and my age is {self.age} ")

    def rest(self):
        print("i am resting")

    @staticmethod
    def printTime():
        print(f" the time is now")

student1= student("ali",12)
student2= student("usman",15)

# print(student1.name)
# print(student1.age)

# print(student2.name)
# print(student2.age)

student1.study()
# student2.study()

student1.rest()
student.printTime()
```

```python
class car :
    def __init__(self,brand,color):
        self.brand=brand
        self.color=color
        
# this is object        
car1= car("Toyota","Black")

print(car1.brand)
print(car1.color)
```

```python
# encapsulation 
class bankaccount:

    def __init__(self,balance):
        self.__balance=balance

    def get_balance(self): # this is getter
        return self.__balance

    def deposit(self,amount): # setter
        if amount >0:
            self.__balance +=amount

a=bankaccount(2000)
a.deposit(500)
print(a.get_balance())
```
