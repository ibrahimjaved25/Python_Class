```python
features = [[150,0],
           [170,1],
           [130,2],
           [160,3],
            [180,4],
            [190,5],
            [200,6],
            [210,7],
            [220,8],
            [230,9],
           [240,10]]
```

```python
labels = ['apple','orange','cheery','watermellon','Mango','Grapes','Strawberry','Banana','Papaya','Pineapple','Guava']
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

```python
prediction=clf.predict([[250, 1]])
```

```python
print(prediction)
```
