# match case in python 

```python
code = 200
```

```python
match code:
    case 200:
        print("Responsive ok")
    case 400:
        print("Bad Response")
    case 404:
        print("not found")
    case _:
        print("Nothing matched ")
print("End of program")
```

```python
Grade = input("Enter your grade ")

match Grade:
    case "A":
        print("Excellent")
    case "B":
        print("Good")
    case "C":
        print("normal")
    case "D":
        print("Bad")
    case _:
        print("Nothing match ")

print(f"your grade is : {Grade}")

```

```python
shop = ["lays","biscuits","chocolates"]
for item in shop:
     if (item == "biscuits"):
        break
     print(item)
```

```python
shop = ["lays","biscuits","chocolates"]
for item in shop:
     if (item == "biscuits"):
        continue # skip this particular iteration of the loop 
     print(item)
```
